```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
labels = ['1920x1080', '2560x1440', '3840x2160']
kornia_times = [142.35, 286.58, 752.17]  # dummy values
image_times = [2212.5, 3940.6, 9051.7]   # dummy values

x = np.arange(len(labels))  # label locations
width = 0.2  # width of the bars

fig, ax = plt.subplots(figsize=(8, 6))

# Plotting the bars
rects1 = ax.bar(x - width/2, kornia_times, width, label='Kornia')
rects2 = ax.bar(x + width/2, image_times, width, label='Image')

# Add labels, title and custom x-axis tick labels
ax.set_xlabel('Resolution')
ax.set_ylabel('Processing Time (µs)')
ax.set_title('Horizontal Flip (u8, RGB)')
ax.set_xticks(x)
ax.set_xticklabels(labels, rotation=30, ha='right')
ax.legend()

# Optional: add value labels on bars
for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f'{height:.1f}',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3),  # offset text upward
                textcoords="offset points",
                ha='center', va='bottom')

plt.tight_layout()
plt.show()
```
